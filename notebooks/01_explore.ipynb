```python
import fastf1
import numpy as np
fastf1.set_log_level('ERROR')
fastf1.Cache.configure(cache_dir='../cache')
```

```python
session = fastf1.get_session(2025, 'Las Vegas', 'R')
session.load(laps=False, telemetry=False, weather=False, messages=False)
x= session.results
```

```python
grid = x['GridPosition'].to_numpy()
finish = x['Position'].to_numpy()
diff = finish-grid
diff_new = [0]*len(grid)
for i in range(len(grid)):
    diff_new[i] = finish[i]-grid[i]

print(diff_new==diff)
```
